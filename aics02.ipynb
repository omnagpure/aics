```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier


# Sample dataset
data = {
    "packet_size": [1000, 1500, 900, 1700, 1100, 1600, 950, 1800],
    "duration": [450, 700, 300, 800, 400, 750, 320, 850],
    "src_packets": [250, 500, 200, 600, 270, 550, 210, 650],
    "dst_packets": [80, 120, 70, 150, 90, 130, 75, 160],
    "label": ["normal", "ddos", "normal", "ddos", "normal", "ddos", "normal", "ddos"]
}

df = pd.DataFrame(data)

# Input and output
X = df[["packet_size", "duration", "src_packets", "dst_packets"]]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = DecisionTreeClassifier()
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

from sklearn.metrics import accuracy_score, confusion_matrix
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# New test data
new_data = [[1200, 800, 300, 130]]
result = model.predict(new_data)

print("Prediction:", result[0])
```
